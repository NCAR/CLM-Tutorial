**Use `run_tower` to create, setup, build and submit a case**

Run_tower handles all four steps of the CESM workflow. 
1. First run_tower creates, sets up, and builds a base case. 
  - This will take a few minutes; then
2. Clones the base case
3. Dowloads the DATM data and initial conditions from a NEON server 
  - This prints a bunch of information to the screen.
  - It also can cause issues if the NEON server gets overloaded or goes down.
4. Submits the {$SITE}.transient case 

After you submit the run_tower script successfully you can check your job status. 

**How do I check my job status?**

You can check job status using the command:
```
qstat -u $USER
```

Case <font face="Courier" color='purple'><strong>setup</strong></font>:
```    
cd /glade/u/home/$USER/cases/ABBY.transient
```

Change the queue to the 'tutorial' queue:
```
./xmlchange JOB_QUEUE=tutorial --force
```

Do not make any other changes since we are running an out-of-the-box simulation.

Build and submit:
```
qcmd -- ./case.build
./case.submit
```
    
Remember that `qcmd` is used on derecho only.

Check the status:
```
qstat -u $USER
```
