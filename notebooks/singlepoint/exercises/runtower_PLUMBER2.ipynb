## Assumptions

Here are some assumptions we've built into NEON simulations that use the *run_tower* script:
- <b>Surface datasets</b>: may have a single plant functional type (PFT), or multiple PFTs, depending on the site, othewise surface datasets have not been modified.
- <b>Data Atmosphere (DATM)</b>: Are of various lengths and set individually for each site.
- <b>Compset</b>: You'll be running a BGC case, but if you run an SP case we have LAI streams for each site.
- <b>Parameter file</b>: are default for the physics version of CLM being used. They **have not** been calibrated to site-level observations  
- <b>Initial conditions</b>: Are not provided, you'll need to do your own spinup for PLUMBER2 cases, therefore the default _run-type_ is for accelerated decomposition (AD) model.

**Use `run_tower` to create, setup, build and submit a case**

Run_tower handles all four steps of the CESM workflow. 
1. First run_tower creates, sets up, and builds a base case. 
  - This will take a few minutes; then
2. Clones the base case
3. Configures the case to use the correct surface dataset and DATM data
  - we have copies of this for PLUMBER2 sites.
4. Submits the {$SITE}.{$EXPERIMENT}.ad case 

After you submit the run_tower script successfully you can check your job status. 

**How do I check my job status?**

You can check job status using the command:
```
qstat -u $USER
```
