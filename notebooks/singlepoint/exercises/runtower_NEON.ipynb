You can find out more information about all of the terrestrial sites at the [NEON science website](https://www.neonscience.org/field-sites)

## Assumptions

Here are some assumptions we've built into NEON simulations that use the *run_tower* script:
- <b>Surface datasets</b>: have a single plant functional type (PFT) that we consider 'dominant' for each NEON tower.  
- <b>Surface datasets</b>: have modified soil properties that are intended to match observations from the NEON MegaPit sampling.
- <b>DATM</b>: are gap filled from NEON and mostly cover 2018-2024, although individual sites may have custom data ranges.
- <b>Compset</b>: You'll be running a BGC case, but if you run an SP case the LAIs **have not** been adjusted with site conditions.
- <b>Parameter file</b>: are default for the physics version of CLM being used. They **have not** been calibrated to site-level observations.  
- <b>Initial conditions</b>: We provide initial conditions that were generated using an AD and postAD spinup for each site. Therefore, the default *run-type* is called 'transient', which means you'll get time-varying CO2, nitrogen deposition, and aerosol deposition (from CMIP6 era forcings). Our aim is to generate new initial condition files that are consistent with the major release version of CLM (e.g. CLM6).  You may need to spinup the model if you're using a different code base, parameter file, surface dataset.

**Use `run_tower` to create, setup, build and submit a case**

Run_tower handles all four steps of the CTSM workflow. 
1. First run_tower creates, sets up, and builds a base case. 
  - This will take a few minutes; then
2. Clones the base case
3. Downloads the DATM data and initial conditions from a NEON server 
  - This prints a bunch of information to the screen.
  - It also can cause issues if the NEON server gets overloaded or goes down.
4. Submits the {$SITE}.transient case 

After you submit the run_tower script successfully you can check your job status. 

**How do I check my job status?**

You can check job status using the command:
```
qstat -u $USER
```
