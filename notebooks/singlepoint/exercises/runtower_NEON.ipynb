## Assumptions

Here are some assumptions we've built into NEON simulations that use the `run_tower` script:
- **Surface datasets**: have a single plant functional type (PFT) that we consider 'dominant' for each NEON tower.  
- **Surface datasets**: have modified soil properties that are intended to match observations from the NEON MegaPit sampling.
- **DATM**: are gap filled from NEON and mostly cover 2018-2024, although individual sites may have custom data ranges.
- **Compset**: You'll be running a BGC case, but if you run an SP case the LAIs **have not** been adjusted with site conditions.
- **Parameter file**: are default for the physics version of CLM being used. They **have not** been calibrated to site-level observations.  
- **Initial conditions**: We provide initial conditions that were generated using an AD and postAD spinup for each site. Therefore, the default *run-type* is called 'transient', which means you'll get time-varying CO2, nitrogen deposition, and aerosol deposition (from CMIP6 era forcings). Our aim is to generate new initial condition files that are consistent with the major release version of CLM (i.e., CLM6). You may need to spinup the model if you're using a different code base, parameter file, or surface dataset. 

## Get started
 
Let's use `run_tower` at a NEON site.

<div class="alert alert-warning">  

**What `run_tower` does**
    
A single `run_tower` command handles the whole CTSM workflow (create, setup, build, and submit). For a NEON site it does four things, in order:

1. **Builds a base case** (e.g., `~/cases/ABBY`). This compiles CTSM once and takes a few minutes.
2. **Clones the base case** into the case you'll actually run (`~/cases/ABBY.transient`). The clone reuses the compiled model, so adding more sites to the `run_tower` command (e.g., `--neon-sites ABBY BART`) is fast.
3. **Downloads the site data**: the atmospheric forcing (DATM) and initial conditions, from NEON's servers. This step prints a lot of output to the screen. If the NEON servers are slow or down, it can fail. You can re-run the same command with `--rerun`.
4. **Submits** the `ABBY.transient` case to the queue.

Once it has been submitted, check on your job with:

```shell
qstat -u $USER
```
