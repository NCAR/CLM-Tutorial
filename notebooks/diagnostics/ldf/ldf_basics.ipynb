# LDF 
**Land Diagnostics Framework**

## Background

The Land Diagnostics Framework (LDF) is based on the Atmosphere Diagnostics Framework (ADF). The LDF currently sits as a branch on the ADF.  Features of this package are:
- Creation of single variables time series and climatologies
- Pairwise comparisons of two land model cases that includes:
  - Summary tables of global means
  - Global timeseries plots of annual means
  - Global Lat-Lon plots of annual and seasonal means
  - Polar view plots of annual and seasonal means
  - Regional climatology and timeseries plots (these include some observations)
- Web site generation

Users can define their test and baseline cases, time ranges, and variables to analyze.

The plotting works on structured (f09) and unstructured (ne30) grids.  At this time both test and baseline cases must be at the same resolution.

### Clone ADF
Navigate to where you want to house the ADF root directory. For this example we will put the ADF in your /work drive on glade. Then clone the code and checkout the branch that we want you to use today (this is a one-time tasks).

```
cd /glade/work/$USER
git clone https://github.com/NCAR/ADF.git
```

This gets the main branch.

Then
```
cd ADF
git fetch origin  # <-- makes other branches available
git checkout -b ldf-native-grid-merge-main origin/ldf-native-grid-merge-main
```

Now you should be on the ldf branch
