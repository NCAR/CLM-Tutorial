# NEON

## CLM-NEON Background

With support from NSF scientists at NCAR and NEON developed new capabilities to easily run Community Land Model (CLM) simulations at NEON tower sites. With this new tool, CLM uses gap-filled meteorology from NEON tower sites, the dominant plant species is mapped to the appropriate model plant functional type (PFT), and soil characteristics used in the simulations are updated to match observations from NEON’s soil megapits. NEON flux measurements are also available for model evaluation.  With the CLM6 release we are providing model configurations and atmospheric forcing for a number of supported towers with data from NEON.  [Lombardozzi et al. (2023)](https://doi.org/10.5194/gmd-16-5979-2023). 

Sample diagnostic here are provided to introduce analyses that can be done with these data products.
