# Function definitions

```python
def ilevel(numlevel):
    return np.arange(numlevel,dtype=float)/float(numlevel-1)
    
def get_history_filenames(idir,htag='h0a',yrs=[]):
    ttag = ''
    if len(yrs) == 2:
        if yrs[0] == yrs[1]:
            ttag = str(yrs[0])
        else:
            ttag = ','.join(str(y) for y in range(yrs[0],yrs[1]+1))
            ttag = '{'+ttag+'}'
            
    command = f'ls {idir}*clm2.{htag}.{ttag}*.nc'
    return subprocess.Popen(command,stdout=subprocess.PIPE,shell='True').communicate()[0].strip().decode().split('\n')
    
def adjust_axes_for_horizontal_colorbar(fig,cb_hgt_in,scale=1.0):
    ''' adjust result of plt.subplots to accomodate a 
    horizontal colorbar at the bottom of the figure'''
    cb_hgt = scale * cb_hgt_in
    ax = fig.get_axes()
    ymin=[]
    ymax=[]
    for a in ax:
        x = a.get_position() # (Bbox defined as [[x0,y0],[x1,y1]])
        ymin = np.append(ymin,np.min([x.y0,x.y1]))
        ymax = np.append(ymax,np.max([x.y0,x.y1]))
        #print ('coords: ', x.x0,x.x1,x.y0,x.y1)
        #print ('width, height: ', x.x1-x.x0,x.y1-x.y0)
    ymin = np.min(ymin)
    ymax = np.max(ymax)
 
    yspan = ymax - ymin ; new_yspan = yspan - cb_hgt
    scale_factor = new_yspan / yspan
    for a in ax:
        x = a.get_position()
        x.y0 = scale_factor*(x.y0 - ymin) + (ymin + cb_hgt)
        x.y1 = scale_factor*(x.y1 - ymin) + (ymin + cb_hgt)
        a.set_position(x)

def setup_cartopy(nrow,ncol,extent):
    fig_proj = ccrs.PlateCarree()
    fig, ax = plt.subplots(nrow,ncol,subplot_kw={'projection': fig_proj})
    fax = fig.get_axes()
    # fill missing data over land
    #landcolor = cmap(0)
    landcolor = 'w'
    for sax in fax:
        sax.add_feature(cfeature.OCEAN,zorder=1.5)
        #sax.add_feature(cfeature.LAND,facecolor=landcolor)
        sax.set_extent(extent, fig_proj)
        sax.coastlines(linewidth=0.5)
        #sax.add_feature(cfeature.COASTLINE)
        gl = sax.gridlines(draw_labels=['left','bottom'])
        gl.xlines,gl.ylines = False,False
        nb = 4
        gl.xlocator = LongitudeLocator(nbins=nb)
        gl.ylocator = LatitudeLocator(nbins=nb)

        gl.xlabel_style = {"size": 8}#, "color": "black"}
        gl.ylabel_style = {"size": 8}
    
    return fig,ax,fax
```

```python
# save all plots to a pdf file
plot_dir = './plots/'
pdf_filename = f'{plot_dir}hillslope_maps.pdf'
SavePlotsPng = False
SavePlotsPng = True

file_prefix = 'map_hillslope'
figcnt = 1

if SavePlotsPng:
    pngfiles = []

RemovePngFiles = False
#RemovePngFiles = True
```

<div class=green16>
Read in CTSM history filenames
</div>

```python
vdir = '/project/tss/swensosc/clm_output/tutorial/western_US_vector_deep_soil_lowland_imped/'
idir = '/project/tss/swensosc/clm_output/tutorial/western_US_gridded_deep_soil_lowland_imped/'

htag1 = 'h1a'
htag0 = 'h0a'
    
yr1 = 1982
yr2 = yr1+0

# 'h1a' column-level files
h1_files = get_history_filenames(idir,htag=htag1,yrs=[yr1,yr2])
# 'h0a' gridcell average files
h0_files = get_history_filenames(vdir,htag=htag0,yrs=[yr1,yr2])

if len(h1_files[0]) < 1:
    print(f'no files found in {idir}')
    exit
```

<div class=purple16>
Define aspects
</div>

```python
aspect_names = ['North','East','South','West']
naspect = len(aspect_names)
aspect_ndx_dict = {aspect_names[n]:(n+1) for n in range(naspect)}
```

<div class=orange16>
    Read history files
</div>

```python
ifile = h1_files[0]
f1 = netcdf4.Dataset(ifile, 'r')
max_columns_hillslope = len(f1.dimensions['max_columns_hillslope']) 
nhillslope = len(f1.dimensions['nhillslope']) 
f1.close()

max_columns_per_hillslope = max_columns_hillslope//nhillslope
nmaxcols = max_columns_per_hillslope # shorter variable name

# plot dimensions
wx,wy = 7,5

# Specify variable to read  
varname,var_units1  = 'H2OSNO','mm'
#varname,var_units1 = 'ZWT','m'
#varname,var_units1 = 'QSOIL','mm/s'
#varname,var_units1 = 'FH2OSFC',''
#varname,var_units1 = 'hillslope_area','m2'

# Convert units from e.g. mm/s to mm/month
convertUnits = False
#convertUnits = True

# Select hillslope (aspect) to show
aspect = 'North'
#aspect = 'South'
hillslope_index = aspect_ndx_dict[aspect]

# Read in all columns from one hillslope
for n in range(nmaxcols):
    hilltag = f'_h{hillslope_index:02d}_c{n+1:02d}'
    hcfiles = [f for f in h1_files if hilltag in f]
    hm = len(hcfiles)
    
    for t in range(hm):
        ifile = hcfiles[t]
        f1 = netcdf4.Dataset(ifile, 'r')

        # Read in coordinates and initialize arrays
        if t==0 and n==0:
            im = len(f1.dimensions['lon']) 
            jm = len(f1.dimensions['lat']) 
            tm = len(f1.dimensions['time']) 
            lon = np.asarray(f1.variables['lon'][:,]) 
            lat = np.asarray(f1.variables['lat'][:,])
            time  = np.zeros((hm))
            pfld1 = np.zeros((nmaxcols,hm,jm,im),dtype=f1.variables[varname].dtype)
            if hasattr(f1.variables[varname],'_FillValue'):
                fill_value = f1.variables[varname].getncattr('_FillValue')
            else:
                fill_value = np.float32(1e36)

        pfld1[n,t,:,:] = np.asarray(f1.variables[varname][:,])
        if n==0: # only need to record once
            time[t] = np.squeeze(f1.variables['time'][:,])
        f1.close()

tm = time.size

# Convert units from mm/s to mm/month
if convertUnits and var_units1=='mm/s':
    cfactor = 30.*24.*3600.
    pfld1[pfld1 < fill_value] = cfactor * pfld1[pfld1 < fill_value]
    var_units1 = 'mm/mo'

pfld1 = np.where(pfld1 >= fill_value,0,pfld1)
```

<div class=gold16>
    Create map of all columns in the specified hillslope
</div>

```python
# set map limits
lonmin = np.min(lon)
lonmax = np.max(lon)
latmin = np.min(lat)
latmax = np.max(lat)

# account for gridcell width
dlon = np.abs(lon[0]-lon[1])
dlat = np.abs(lat[0]-lat[1])
lonmin -= 0.5*dlon
lonmax += 0.5*dlon
latmin -= 0.5*dlat
latmax += 0.5*dlat
        
# Calculate time average fields
meanvar1 = np.zeros((nmaxcols,jm,im))
tmp  = np.zeros((tm,jm,im))
# some gridcells may not have all nmaxcols columns
# so filter for valid columns
for n in range(nmaxcols):
    tmp[:,] = np.copy(pfld1[n,:,])
    cnt = np.ones((tm,jm,im))
    ind=np.where(tmp >= fill_value)
    tmp[ind] = 0
    cnt[ind] = 0
    norm = np.sum(cnt,axis=0) ; norm[norm == 0] = 1
    meanvar1[n,:,] = np.sum(tmp,axis=0)/norm

#meanvar1 = np.sum(pfld1,axis=1)/float(tm)

# plot contours or raster
useContour = False
useContour = True

xmn = np.argmin(np.abs(lonmin-lon))
xmx = np.argmin(np.abs(lonmax-lon))
ymn = np.argmin(np.abs(latmin-lat))
ymx = np.argmin(np.abs(latmax-lat))

# create west longitude array
lonw = np.copy(lon)
lonw[lonw>180] -= 360

if useContour:
    extent = [lonw[xmn],lonw[xmx],lat[ymn],lat[ymx]]
else:
    extent = [lonw[xmn]-0.5*dlon,lonw[xmx]+0.5*dlon,lat[ymn]-0.5*dlat,lat[ymx]+0.5*dlat]

nrow = 2
ncol = nmaxcols//nrow
if useCartopy:
    fig2,ax,fax = setup_cartopy(nrow,ncol,extent)
else:
    fig2, ax = plt.subplots(nrow,ncol)
    fax = fig2.get_axes()
        
bbox1 = fax[0].get_position() 
bbox2 = fax[ncol-1].get_position() 
        
# modify bounding box of axes to accomodate colorbar at bottom of plot 
cb_hgt = 0.03
adjust_axes_for_horizontal_colorbar(fig2,cb_hgt,scale=2.)

# add the colorbar axes
cb_w = (bbox2.x1 - bbox1.x0)

cbaxes  = fig2.add_axes([bbox1.x0+0.1*cb_w,0.06,0.8*cb_w,cb_hgt])

fig2.set_size_inches([wx,wy])

# set figure title
plot_title = f'Mean {varname} Hillslope {aspect}'
fig2.suptitle(plot_title,fontsize=16)

for sax in fax:
    sax.tick_params(labelsize=8)

lnum=25
# set contour levels
if 1==2:
    minfld = np.min(meanvar1)
    maxfld = np.max(meanvar1)
else:
    fld_sorted = np.sort(np.abs(meanvar1.flat))
    f = 0.95
    minfld = fld_sorted[int((1-f)*fld_sorted.size-1)]
    maxfld = fld_sorted[int((f)*fld_sorted.size-1)]

# set range manually
if varname == 'ZWT':
    minfld,maxfld = 0,8
if varname == 'H2OSNO':
    minfld,maxfld = 0,200
    
datarange = maxfld - minfld
level = datarange*ilevel(lnum) + minfld
cblevel = datarange*ilevel(5) + minfld
cnorm = matplotlib.colors.BoundaryNorm(level, 256)

#cmap = plt.cm.jet
cmap = plt.cm.gist_earth#_r

if 1==1:
    img = []
    for n in range(nmaxcols):
        colmask = np.where(meanvar1[n,] >= fill_value,1,0)
        tmp = np.ma.masked_array(meanvar1[n,ymn:ymx+1,xmn:xmx+1],mask=colmask[ymn:ymx+1,xmn:xmx+1])
        if useContour:
            img.append(fax[n].contourf(lonw[xmn:xmx+1],lat[ymn:ymx+1],tmp,norm=cnorm,levels=level,cmap=cmap))
        else:
            img.append(fax[n].imshow(np.flipud(tmp),norm=cnorm,extent=extent,cmap=cmap,aspect='auto'))
        fax[n].set_title(f'Col {n+1}')

    cbar = fig2.colorbar(img[0],ax=fax[0],ticks=cblevel,label=var_units1,
                         orientation="horizontal",cax = cbaxes)
    
if SavePlotsPng:
    fname = '{}{}.fig_{:d}.png'.format(plot_dir,file_prefix,figcnt)
    pngfiles.append(fname)
    plt.savefig(fname,format='png')
    figcnt+=1
```

<div class="alert alert-info">
<strong>What attribute (geomorphic parameter) could explain the different snowpack magnitudes?</strong>

<div class="alert alert-success">
<details>
<summary>Answer</summary>

Elevation.  

The columns are ordered according to elevation.  Column 3 is the highest elevation column, and has colder temperatures due to the elevation-based downscaling of temperature.

</details>

<div class="alert alert-info">
<strong>Why do columns 2-4 have smaller snowpack magnitudes than column 1 (the lowland column) in some locations?</strong>

<div class="alert alert-success">
<details>
<summary>Answer</summary>

Surface water.  

Subsurface lateral flow causes higher soil moisture values in the lowland column, which may result in the presence of surface water.  As surface water freezes, it is transferred to the snowpack.

</details>

<div class=purple16>
Compare hillslope averages
</div>

```python
# read in column (area) weights
weightfile = '/project/tss/swensosc/clm_output/tutorial/western_US_gridded/hillslope_western_US_hillslope_geometry.nc'

f1 = netcdf4.Dataset(weightfile, 'r')
wtlun = np.asarray(f1.variables['cols1d_wtlunit'][:,])
wtlun[wtlun >= f1.variables['cols1d_wtlunit'].getncattr('_FillValue')] = 0
f1.close()

hillslope_mean = np.zeros((nhillslope,tm,jm,im))
norm = np.zeros((nhillslope,jm,im))
# loop over all columns of all hillslopes
for nhill in range(nhillslope):
    for n in range(nmaxcols):
        hilltag = f'_h{nhill+1:02d}_c{n+1:02d}'
        hcfiles = [f for f in h1_files if hilltag in f]
        hm = len(hcfiles)
        for t in range(hm):
            ifile = hcfiles[t]

            f1 = netcdf4.Dataset(ifile, 'r')
            nh = f1.variables[varname].hillslope_index
            nc = f1.variables[varname].column_position
            tmp = np.asarray(f1.variables[varname][0,])
            tmp[tmp >= f1.variables[varname].getncattr('_FillValue')] = 0
            hillslope_mean[nhill,t,:,:] += wtlun[nh-1,nc-1,]*tmp
            if t==0:
                norm[nhill,:,:] += wtlun[nh-1,nc-1,]
            f1.close()

# normalize weights
for t in range(hm):
    tmp = hillslope_mean[:,t,:,:]
    tmp[norm>0] = tmp[norm>0]/norm[norm>0]
    hillslope_mean[:,t,:,:] = tmp

nrow = 2
ncol = nmaxcols//nrow
if useCartopy:
    fig2,ax,fax = setup_cartopy(nrow,ncol,extent)
else:
    fig2, ax = plt.subplots(nrow,ncol)
    fax = fig2.get_axes()
        
bbox1 = fax[0].get_position() 
bbox2 = fax[ncol-1].get_position() 
        
# modify bounding box of axes to accomodate colorbar at bottom of plot 
cb_hgt = 0.03
adjust_axes_for_horizontal_colorbar(fig2,cb_hgt,scale=2.)

# add the colorbar axes
cb_w = (bbox2.x1 - bbox1.x0)

cbaxes  = fig2.add_axes([bbox1.x0+0.1*cb_w,0.06,0.8*cb_w,cb_hgt])

fig2.set_size_inches([wx,wy])

# set figure title
plot_title = f'Mean {varname} Hillslope Average'
fig2.suptitle(plot_title,fontsize=16)

for sax in fax:
    sax.tick_params(labelsize=8)

lnum = 25
# set contour levels
if 1==2:
    minfld = np.min(meanvar1)
    maxfld = np.max(meanvar1)
else:
    fld_sorted = np.sort(np.abs(meanvar1.flat))
    f = 0.95
    minfld = fld_sorted[int((1-f)*fld_sorted.size-1)]
    maxfld = fld_sorted[int((f)*fld_sorted.size-1)]

# set range manually
if varname == 'ZWT':
    minfld,maxfld = 0,8
if varname == 'H2OSNO':
    minfld,maxfld = 0,100
    
datarange = maxfld - minfld
level = datarange*ilevel(lnum) + minfld
cblevel = datarange*ilevel(5) + minfld
cnorm = matplotlib.colors.BoundaryNorm(level, 256)

#cmap = plt.cm.jet
cmap = plt.cm.gist_earth#_r

if 1==1:
    img = []
    for n in range(nhillslope):
        colmask = np.where(meanvar1[n,] >= fill_value,1,0)
        tmp = np.ma.masked_array(np.mean(hillslope_mean[n,:,ymn:ymx+1,xmn:xmx+1],axis=0),mask=colmask[ymn:ymx+1,xmn:xmx+1])
        #tmp = np.ma.masked_array(norm[n,ymn:ymx+1,xmn:xmx+1],mask=colmask[ymn:ymx+1,xmn:xmx+1])
        if useContour:
            img.append(fax[n].contourf(lonw[xmn:xmx+1],lat[ymn:ymx+1],tmp,norm=cnorm,levels=level,cmap=cmap))
        else:
            img.append(fax[n].imshow(np.flipud(tmp),norm=cnorm,extent=extent,cmap=cmap,aspect='auto'))
        fax[n].set_title(f'{aspect_names[n]}')

    cbar = fig2.colorbar(img[0],ax=fax[0],ticks=cblevel,label=var_units1,
                         orientation="horizontal",cax = cbaxes)
    
if SavePlotsPng:
    fname = '{}{}.fig_{:d}.png'.format(plot_dir,file_prefix,figcnt)
    pngfiles.append(fname)
    plt.savefig(fname,format='png')
    figcnt+=1
```

```python
# read in column weights
weightfile = '/project/tss/swensosc/clm_output/tutorial/western_US_gridded/hillslope_western_US_hillslope_geometry.nc'

f1 = netcdf4.Dataset(weightfile, 'r')
# recall: hillslope gridcell weights will not sum to 1 if other landunits are present in the gridcell
wtgrc = np.asarray(f1.variables['cols1d_wtgcell'][:,])
wtgrc[wtgrc >= f1.variables['cols1d_wtgcell'].getncattr('_FillValue')] = 0
f1.close()

grc_mean = np.zeros((tm,jm,im))
# loop over all columns of all hillslopes to get gridcell mean
for nhill in range(nhillslope):
    for n in range(nmaxcols):
        hilltag = f'_h{nhill+1:02d}_c{n+1:02d}'
        hcfiles = [f for f in h1_files if hilltag in f]
        hm = len(hcfiles)
        for t in range(hm):
            ifile = hcfiles[t]

            f1 = netcdf4.Dataset(ifile, 'r')
            nh = f1.variables[varname].hillslope_index
            nc = f1.variables[varname].column_position
            tmp = np.asarray(f1.variables[varname][0,])
            tmp[tmp >= f1.variables[varname].getncattr('_FillValue')] = 0
            grc_mean[t,:,:] += wtgrc[nh-1,nc-1,]*tmp
            f1.close()

# Read in gridcell-level history output
grc_h0 = np.zeros((tm,jm,im))
for t in range(tm):
    ifile = h0_files[t]
    f1 = netcdf4.Dataset(ifile, 'r')
    tmp = np.asarray(f1.variables[varname][0,])
    tmp[tmp >= f1.variables[varname].getncattr('_FillValue')] = 0
    grc_h0[t,:,:] = tmp
    f1.close()
    
nrow = 3
ncol = 1
if useCartopy:
    fig2,ax,fax = setup_cartopy(nrow,ncol,extent)
else:
    fig2, ax = plt.subplots(nrow,ncol)
    fax = fig2.get_axes()

fig2.subplots_adjust(hspace=0.5)

wx,wy = 5,6
#fig2.set_size_inches([wx,wy])

# set figure title
plot_title = f'Mean annual {varname} Gridcell Average'
fig2.suptitle(plot_title,fontsize=16)

xmn = np.argmin(np.abs(lonmin-lon))
xmx = np.argmin(np.abs(lonmax-lon))
ymn = np.argmin(np.abs(latmin-lat))
ymx = np.argmin(np.abs(latmax-lat))

for sax in fax:
    sax.tick_params(labelsize=10)

# set contour levels
lnum = 25
if 1==2:
    minfld = 0
    maxfld = np.max(grc_h0)
else:
    fld_sorted = np.sort(np.abs(grc_h0.flat))
    f = 0.95
    minfld = fld_sorted[int((1-f)*fld_sorted.size-1)]
    maxfld = fld_sorted[int((f)*fld_sorted.size-1)]

datarange = maxfld - minfld
level = datarange*ilevel(lnum) + minfld
cblevel = datarange*ilevel(5) + minfld
cnorm = matplotlib.colors.BoundaryNorm(level, 256)

cmap  = plt.cm.gist_earth#_r
cmap2 = plt.cm.bwr

tmp = np.mean(grc_mean[:,ymn:ymx+1,xmn:xmx+1],axis=0)
if useContour:
    img1 = fax[0].contourf(lonw[xmn:xmx+1],lat[ymn:ymx+1],tmp,norm=cnorm,levels=level,cmap=cmap)
else:
    img1 = fax[0].imshow(np.flipud(tmp),norm=cnorm,extent=extent,cmap=cmap,aspect='auto')

tmp = np.mean(grc_h0[:,ymn:ymx+1,xmn:xmx+1],axis=0)
if useContour:
    img2 = fax[1].contourf(lonw[xmn:xmx+1],lat[ymn:ymx+1],tmp,norm=cnorm,levels=level,cmap=cmap)
else:
    img2 = fax[1].imshow(np.flipud(tmp),norm=cnorm,extent=extent,cmap=cmap,aspect='auto')

cbar1 = fig2.colorbar(img1,ax=fax[0],ticks=cblevel,label=var_units1,
                     orientation="vertical")
cbar2 = fig2.colorbar(img2,ax=fax[1],ticks=cblevel,label=var_units1,
                     orientation="vertical")

maxfld2 = 0.1 * maxfld
minfld2 = -maxfld2
datarange2 = maxfld2 - minfld2
level2 = datarange2*ilevel(lnum) + minfld2
cblevel2 = datarange2*ilevel(5) + minfld2
cnorm2 = matplotlib.colors.BoundaryNorm(level2, 256)

tmp = np.mean(grc_h0[:,ymn:ymx+1,xmn:xmx+1],axis=0) - np.mean(grc_mean[:,ymn:ymx+1,xmn:xmx+1],axis=0)
if useContour:
    img3 = fax[2].contourf(lonw[xmn:xmx+1],lat[ymn:ymx+1],tmp,norm=cnorm2,levels=level2,cmap=cmap2)
else:
    img3 = fax[2].imshow(np.flipud(tmp),norm=cnorm2,extent=extent,cmap=cmap2,aspect='auto')

cbar3 = fig2.colorbar(img3,ax=fax[2],ticks=cblevel2,label=var_units1,
                     orientation="vertical")

fax[0].set_title('sum hillslope columns')
fax[1].set_title('gridcell average')
fax[2].set_title('difference')

if SavePlotsPng:
    fname = '{}{}.fig_{:d}.png'.format(plot_dir,file_prefix,figcnt)
    pngfiles.append(fname)
    plt.savefig(fname,format='png')
    figcnt+=1
```

```python
yr1,yr2 = 1981,1984

# h0/h1 are gridded/vector history files, respectively
h0files = get_history_filenames(vdir,htag=htag0,yrs=[yr1,yr2])
h1files = get_history_filenames(vdir,htag=htag1,yrs=[yr1,yr2])

hm = len(h1files)
if len(h1files[0]) < 1:
    print('no files found in '+idir)
    stop

# get hillslope dimensions from 1st vector file
ifile = h1files[0]
f1 = netcdf4.Dataset(ifile, 'r')
max_columns_hillslope = len(f1.dimensions['max_columns_hillslope']) 
nhillslope = len(f1.dimensions['nhillslope']) 
f1.close()

max_columns_per_hillslope = max_columns_hillslope//nhillslope
nmaxcols = max_columns_per_hillslope

# Specify point
ptnum = 1
if ptnum == 1: #
    plon = 245
    plat = 46
if ptnum == 2: #
    plon = 250
    plat = 46
if ptnum == 3: #
    plon = 250
    plat = 35

pt = [plon,plat]

# read in data
f1 = netcdf4.MFDataset(h0files, 'r')
im  = len(f1.dimensions['lon']) 
jm  = len(f1.dimensions['lat']) 
tm  = len(f1.dimensions['time']) 
lon = f1.variables['lon'][:,] 
lat = f1.variables['lat'][:,]
pt_i, pt_j = np.argmin(np.abs(plon-lon)),np.argmin(np.abs(plat-lat))
slon, slat = lon[pt_i],lat[pt_j]
ctsm_ref_year = int(f1.variables['time'].units.split()[2].split('-')[0])
time = f1.variables['time'][:,]

if 'VOLUMETRIC_STREAMFLOW' in f1.variables.keys():
    streamflow = np.asarray(f1.variables['VOLUMETRIC_STREAMFLOW'][:,pt_j,pt_i])
if 'STREAM_WATER_DEPTH' in f1.variables.keys():
    stream_water_depth = np.asarray(f1.variables['STREAM_WATER_DEPTH'][:,pt_j,pt_i])
if 'STREAM_WATER_VOLUME' in f1.variables.keys():
    stream_water_vol = np.asarray(f1.variables['STREAM_WATER_VOLUME'][:,pt_j,pt_i])
    
f1.close()

# adjust point to ctsm grid
plon = lon[pt_i]
plat = lat[pt_j]
pt = [plon,plat]

# read soil structure variables
f1 = netcdf4.Dataset(get_history_filenames(vdir,htag=htag0)[0], 'r')
nlevsoi = len(f1.dimensions['levsoi'])
zfv = f1.variables['ZSOI']._FillValue
zsoi = np.asarray(f1.variables['ZSOI'][:,pt_j,pt_i])
dzsoi = np.asarray(f1.variables['DZSOI'][:,pt_j,pt_i])
watsat = np.asarray(f1.variables['WATSAT'][:,pt_j,pt_i])
pct_sand = np.asarray(f1.variables['PCT_SAND'][:,pt_j,pt_i])
pct_clay = np.asarray(f1.variables['PCT_CLAY'][:,pt_j,pt_i])
f1.close()
# excluding surface (zisoi = 0)
zisoi = zsoi + 0.5*dzsoi

# hillslope data
hprefix = 'hillslope'
f1 = netcdf4.Dataset(h1files[0], 'r')
lon1d = f1.variables['cols1d_lon'][:,]
lat1d = f1.variables['cols1d_lat'][:,]
hillslope_index = f1.variables[f'{hprefix}_index'][:,]
eps = 1e-5
l1 = (np.abs(slon-lon1d) < eps)
l2 = (np.abs(slat-lat1d) < eps)

# first filter for hillslope columns at specified location
hind = np.logical_and.reduce((l1,l2,hillslope_index>0))
# select single hillslope
# north, east, south, west [1-4]
nasp = 1
# now filter for specified aspect
hind = np.logical_and(hillslope_index==nasp,hind)
    
hillslope_index =  f1.variables[f'{hprefix}_index'][...,hind]
hillslope_elev =  f1.variables[f'{hprefix}_elev'][...,hind]
hillslope_dist =  f1.variables[f'{hprefix}_distance'][...,hind]
hillslope_area =  f1.variables[f'{hprefix}_area'][...,hind]
nbedrock =  f1.variables['cols1d_nbedrock'][...,hind]
zbedrock = zisoi[nbedrock]
f1.close()

# single point data
f1 = netcdf4.MFDataset(h1files, 'r')
zwt = f1.variables['ZWT'][...,hind]
sm  = f1.variables['SOILLIQ'][...,hind]
sm += f1.variables['SOILICE'][...,hind]

frost_table = f1.variables['FROST_TABLE'][...,hind]

fh2osfc = f1.variables['FH2OSFC'][...,hind]
h2osfc  = f1.variables['H2OSFC'][...,hind]
fsno    = f1.variables['FSNO'][...,hind]
h2osno  = f1.variables['H2OSNO'][...,hind]

qlatflowout = f1.variables['QLATFLOWOUT'][...,hind]
qdrai = f1.variables['QDRAI'][...,hind]
qover = f1.variables['QOVER'][...,hind]
qvegt = f1.variables['QVEGT'][...,hind]
qvege = f1.variables['QVEGE'][...,hind]
qsoil = f1.variables['QSOIL'][...,hind]

f1.close()

#calcSaturation = False
calcSaturation = True

if calcSaturation:
    # convert vwc to saturation (0-1)
    sm_units = '' 
    for k in range(nlevsoi):      
        sm[:,k,:] = sm[:,k,:]/(1.e3 * dzsoi[k] * watsat[k])
else:
    sm_units = 'm3/m3' 
    for k in range(nlevsoi):      
        sm[:,k,:] = sm[:,k,:]/(1.e3 * dzsoi[k])

ytime = time/365+ctsm_ref_year
```
