# Exercise Overview

## Learning Goals

In this exercise, you will gain hands-on experience modifying namelists and examining the impact of those changes on a simulation.

By the end of this exercise, you will be able to:

- Edit the land `user_nl_clm` file.
- Generate and inspect updated namelists.
- Run a CLM simulation with custom namelist settings.
- Verify that your namelist changes have been applied correctly.
