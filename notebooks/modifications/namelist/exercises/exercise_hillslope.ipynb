## Exercise: Set namelist options for a hillslope simulation

A hillslope hydrology simulation is activated via namelist variables in `user_nl_clm`:

<p>
    
```fortran
hillslope_file = '/glade/p/cesmdata/cseg/inputdata/lnd/clm2/hillslopedata/hillslopes_0.9x1.25_c240416.nc'
```

</p>

The hillslope file contains information on the hillslope geomorphic parameters.  Optionally, it can specify soil thickness (depth to bedrock) and a pft for each hillslope column.  

<p>
    
```fortran
use_hillslope = .true.
use_hillslope_routing = .true.
hillslope_fsat_equals_zero = .true.
```

</p>

<p>
    
`use_hillslope` activates the hillslope hydrology configuration.

`use_hillslope_routing` specifies that the internal hillslope stream channel parameterization will be used as the lower boundary condition (recommended).  

`hillslope_fsat_equals_zero` disables the TOPMODEL-based surface runoff approach (recommended).

</p>

<h2>
Obtaining column-level history output
</h2>

To add a second history file containing column-level output, set `hist_dov2xy = .false.` and specify which fields to output:

```fortran
hist_nhtfrq = 0,0
hist_mfilt = 1,1
hist_dov2xy = .true., .false.

! 'COLS' will average pft variables to column level for vector output
hist_type1d_pertape = 'GRID','COLS'

hist_fincl1 = 'STREAM_WATER_DEPTH'

hist_fincl2 ='H2OSFC','H2OSNO','ZWT','ZWT_PERCH','FROST_TABLE','TWS','SOILLIQ','SOILICE',
'FSNO','FH2OSFC','QLATFLOWOUT','QDRAI','QDRAI_PERCH','QOVER','QVEGT','QVEGE','QSOIL','TSOI'
```

<h2>
Methods for distributing pfts across hillslope columns
</h2>

<p>
    
```fortran
pft_distribution_method 
```

</p>

Options for `pft_distribution_method`

`'Standard'`: Identical pfts on each column

`'DominantPftUniform'`: Superceded by `n_dom_pfts = 1`

`'DominantPftLowland'`: Largest pft assigned to lowland, 2nd largest to uplands

`'PftLowlandUpland'`: Directly assign lowland/upland pfts

Pft indices can be specified in the source code in `src/biogeophys/HillslopeHydrologyMod.F90`

Example: `call HillslopeSetLowlandUplandPfts(bounds,lowland_ivt=7,upland_ivt=13)` sets the lowland column pft to 7 (broadleaf deciduous temperate tree) and the upland columns to 13 (C3 non-arctic grass)

`'FromFile'`: Assign pfts for each column on input hillslope file. (This option requires appropriate data on the hillslope input file)

<h2>
Methods for varying soil thickness across hillslope columns
</h2>

<p>
    
```fortran
soil_profile_method
```

</p>

Options for `soil_profile_method`

`'Uniform'`: Identical values for all columns

`'SetLowlandUpland'`: Directly assign lowland/upland soil thickness

Lowland and upland soil thicknesses can be specified in the source code in `src/main/clm_instMod.F90`

Example: `call SetHillslopeSoilThickness(bounds, hillslope_file, soil_depth_lowland_in=8.5_r8,soil_depth_upland_in=2.0_r8)`    
            sets the lowland column soil thickness to 8.5m and the upland columns to 2m.

`'Linear'`: Assign lowland/upland soil thickness with linear ramp

`'FromFile'`: Assign soil thickness for each column on input hillslope file.  (This option requires appropriate data on the hillslope input file)
