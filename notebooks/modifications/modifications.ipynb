# Modifying Your Case

So far you have learned how to create and run basic CLM and FATES cases. However, we often want to customize a case: turning specific options on or off, changing the history output, or using different input files.

There are two main ways to modify a case, and this chapter covers both:

- **XML Modifications:** XML files control case-level settings such as run length, run type, and model time step.
- **Namelist Modifications:** Namelists control component-specific settings such as history output, parameter files, and physics options.
