## Exercise: Create a hybrid case 
    
Create a hybrid case from another run to create a new case named "i2000_hybrid_case".  

For our reference case, we will use the run you did in the [Change component set chapter](../../../basics/exercises/change_compsets.ipynb), e.g., ``i2000_fates_sp.basics`` at month 1. The restart files for this run are located in: 

```
/glade/derecho/scratch/$USER/archive/i2000_fates_sp.basics/rest/2000-02-01-00000
```
Change the ``i2000_fates_sp.basics`` in the path to whatever you named your case, and possibly the last folder to grab the last timestep.

Run the new hybrid case for 1 month, then resubmit for 1 month.

1. Create your new case with the command:
    
```
    cd /glade/u/home/$USER/code/my_cesm_code/cime/scripts/
    ./create_newcase --case /glade/u/home/$USER/cases/i2000_hybrid_case --res f19_f19_mt233 --compset I1850Clm60SpCrujra --run-unsupported
```

2. In your case directory, change the run type to hybrid, and change the associated RUN_* variables:

```
    cd /glade/u/home/$USER/cases/i2000_hybrid_case
    ./xmlchange RUN_TYPE=hybrid
    ./xmlchange RUN_REFCASE=i2000_fates_sp.basics
    ./xmlchange GET_REFCASE=True
    ./xmlchange RUN_REFDIR=/glade/derecho/scratch/$USER/archive/i2000_fates_sp.basics/rest/2000-02-01-00000
    ./xmlchange RUN_REFDATE=2000-02-01-00000
```

3. Set run time variables: 
    
```
    ./xmlchange STOP_OPTION=nmonths
    ./xmlchange STOP_N=1
    ./xmlchange RESUBMIT=1
    ./xmlchange JOB_WALLCLOCK_TIME=02:00:00
    ./xmlchange JOB_QUEUE=tutorial --force
```    

4. Set up the case:
    
```
    ./case.setup
```
    
5. Run ./preview_namelists:

```
    ./preview_namelists
```
    
6. Carry on to build and submit the run:
    
```
      qcmd -- ./case.build
      ./case.submit
```

Remember that `qcmd` is used on Derecho only.
    
After the job completes, go to the short term archive space and explore.

</details>
</div>
