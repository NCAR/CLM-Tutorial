Then use ``xmlchange`` to make the model cycle over a different block of forcing years. Set ``DATM_YR_START`` to 2005 and ``DATM_YR_END`` to 2009. Set ``DATM_YR_ALIGN`` to the year your run starts (the year in ``RUN_STARTDATE``), so that the first model year uses the 2005 forcing.

The data atmosphere (DATM) reads a fixed block of forcing years and cycles through them over the course of your run. Three variables control this:

- ``DATM_YR_START``: the first year of forcing data to use.
- ``DATM_YR_END``: the last year of forcing data to use.
- ``DATM_YR_ALIGN``: the model year that is matched to ``DATM_YR_START``.

For example, for a run that starts in model year 2000 with ``DATM_YR_START=2005``, ``DATM_YR_END=2009``, and ``DATM_YR_ALIGN=2000``, model year 2000 is forced with 2005 data, model year 2001 with 2006 data, and so on up to model year 2004 with 2009 data. In model year 2005 the forcing wraps back around to 2005.

1. From the script directory, create your case:
    
```
    cd /glade/u/home/$USER/code/my_cesm_code/cime/scripts/
    ./create_newcase --case /glade/u/home/$USER/cases/i2000_forcing_years --res f19_f19_mt233 --compset I1850Clm60SpCrujra --run-unsupported
```
    
2. In your case directory, change the runtime variables:
    
```
    cd /glade/u/home/$USER/cases/i2000_forcing_years
    ./xmlchange DATM_YR_START=2005
    ./xmlchange DATM_YR_END=2009
    ./xmlchange DATM_YR_ALIGN=2000
    ./xmlchange STOP_N=1
    ./xmlchange STOP_OPTION='nyears'
    ./xmlchange JOB_WALLCLOCK_TIME=1:00:00
    ./xmlchange JOB_QUEUE=tutorial --force
``` 

3. Set up the case:
    
```
    ./case.setup
```
    

4. Continue to build and submit: 
    
```
    qcmd -- ./case.build
    ./case.submit
```
    
Remember that `qcmd` is used on Derecho only.

</details>
</div>
