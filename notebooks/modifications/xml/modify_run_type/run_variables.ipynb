# Variables Related to Run Type

A summary of the xml variables pertaining to run type in `env_run.xml`. 

<br>

***


| Variable | Run Type | Description | Valid values / notes |
|---|---|---|---|
| `RUN_TYPE` | | Determines the model run initialization type. Only important for the initial run of a production run when `$CONTINUE_RUN` is set to `FALSE`. | `'startup'`, `'hybrid'`, `'branch'` |
| `RUN_STARTDATE` | Startup or hybrid | Run start date. | `yyyy-mm-dd` |
| `RUN_REFCASE` | Hybrid or branch | The name of the reference case you are starting from. | Case name |
| `RUN_REFDATE` | Hybrid or branch | The date stamp of the reference case you are starting from. | `yyyy-mm-dd` |
| `DRV_RESTART_POINTER` | Branch | The name of the coupler's restart pointer file of the reference case you are starting from. You must append the restart date and time to `rpointer.cpl`. *(NEW in CESM3)* | Append `yyyy-mm-dd-sssss` to `rpointer.cpl` |
| `GET_REFCASE` | Hybrid or branch | Flag for automatically prestaging the refcase restart dataset. If `TRUE`, the refcase data is prestaged into the executable directory. If `FALSE`, you need to manually put the restart files into `$RUNDIR`. | `TRUE` / `FALSE` |
