## Step 1: Create CLM Code Directory


First, create a location in the **glade** file system where you will store your CLM source code.

Throughout this tutorial, `$USER` is the environment variable already set for you that is your NCAR HPC username.

<div class="alert alert-info"> 
<strong>Exercise: Create a directory to store your code </strong><br>
NOTE: You may have already done this step.

```
cd /glade/u/home/$USER
mkdir code
``` 
</div>

## Step 2. Clone the CTSM Repository

First, we will change into the `code` workspace location then we will use the `git clone` command to download the CLM code.

<div class="alert alert-info">
<strong>Exercise: Clone the CTSM Repository </strong><br>
    
- **Change the current directory to the code workspace directory**: 

```
    cd /glade/u/home/$USER/code
```
<br>
    
- **Download the CLM code to your code workspace directory as `my_clm_code`:**

```
    git clone https://github.com/ESCOMP/CTSM.git my_clm_code
    cd my_clm_code
    git checkout ctsm5.4.063
```
The directory name `my_clm_code` is only an example. You may choose any directory name you prefer. Outside of this tutorial, many users choose names that identify the CLM version they are working with (for example, `ctsm6.0`, `ctsm5.5` or `clm6.0_Tutorial2026`).

- **Verify the Tutorial Tag**

You can verify that you are using the correct version by running:

```
    git describe --tags --exact-match
```

which should return:

```
    ctsm5.4.063
```
</div>

### Versions

**CLM Versions**

One of the first decisions when starting a modeling project is choosing which version of CLM to use. Factors to consider include:

- Scientific and technical support for the version. As a rule of thumb, release tags are the most supported version of CLM. 
- Availability of model features required for your research.
- Consistency with community experiments or previous studies.
- Compatibility with analysis tools and workflows.

For example, if you plan to compare your simulations with a community project, you should generally use the same CESM or CLM version as that project.

More information can be found on the [CLM github Page](https://github.com/ESCOMP/CTSM.git).


<div class="alert alert-info">
    
To see available CLM tags that have a release tag (and thus a DOI to go with them) go to the release page on CLM github. It's under the Code tab on the far left of the page, and then "Tags" two to the right of the "master" pulldown, and then Releases. (https://github.com/ESCOMP/CTSM/releases)

</div>


## Step 3. Download Component Models with `git-fleximod` 

The CTSM repository itself does not contain all component model source code. **Individual model components are maintained in separate repositories** and must be downloaded after cloning CLM. 

CESM3/CLM uses `git-fleximod` to manage external repositories. 

From your CLM source directory, run: 

```bash
    cd /glade/u/home/$USER/code/my_clm_code 
    ./bin/git-fleximod update
```

Depending on your system configuration, you may be prompted to accept a security certificate. If prompted, accept the certificate permanently. If no prompt appears, that is perfectly normal. During this step, `git-fleximod` will download the required component models and supporting repositories.

## Step 4. Set up a `ctsm_pylib` conda environment 

Conda environments provide a separate, self-contained description of Python packages that can be used for different tasks. Each environment is isolated from the others and from your system Python. Conda environments allow for reproducibility, testing, and stable Python environments for scientific workflows.  We created the ctsm_pylib environment for various python tools we'll use in CLM.

From the top-level of your CLM checkout
```cd /glade/u/home/$USER/code/my_clm_code```

you can simply run:
```./py_env_create```

Heads up, this can take a while.
