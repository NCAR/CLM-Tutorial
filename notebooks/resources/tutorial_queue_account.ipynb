# Tutorial queue and account

## Tutorial Project Account

<div class="alert alert-warning">

You should have access to project account {{ PROJECT_CODE }} and use this for your simulations.

</div>

You can change the account in your case directory using the following command:

``` {parsed-literal}
./xmlchange PROJECT={{ '`' + PROJECT_CODE + '`' }}
```

## Special Queues

<div class="alert alert-warning">

**Tutorial Reservation Queues**
    
- We have a special queue called **{{ TUTORIAL_QUE }}** during the lab sessions to ensure you get through the derecho queues quickly and get your jobs run. This queue is only active during the day. 

- Outside the lab sessions, you should use the queue **main** to run CLM.
    
</div>

If you need to change the  value for a run, you can change the reservation queue in your case directory using the following command:

During the lab sessions:

``` {parsed-literal}
./xmlchange JOB_QUEUE={{ '`' + TUTORIAL_QUE + '`' }} --force
```

Outside the lab sessions:

``` bash
./xmlchange JOB_QUEUE=main
```

## Compiling in the Special Queue

By default, we compile CTSM on a shared queue called `develop`. However, as we want to avoid long wait time during the labs, we will compile in the special reservation queue {{ TUTORIAL_QUE }}.

During the lab sessions, use

``` bash
qcmd -q tutorial -- ./case.build 
```


Outside the lab sessions, use

``` bash
qcmd -- ./case.build 
```
