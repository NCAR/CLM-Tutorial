## Step 4. Examine Instantaneous Snow Persistence

Components in CESM generally seperate history fields into seperate files by their averaging type and timescale. In the land model, there are two history files by default. The "h0a" file holds fields averaged for the time period specified in the filename and file metadata. The "h0i" file hold instantaneous data from the timestep shown in the filename and the file metadata. In our basic example, the only field in the instantaneous file is for snow persistence "SNOW_PERSISTENCE".


<strong>Open the h0i file using `ncvis`: </strong><br>
```
ncvis i1850_sp.basics.clm2.h0i.0001-01.nc &
```

<div class="alert alert-info">

<strong>Select the snow persistence variable: </strong><br><br>

Click on the **(1) 3D vars** button. Date written and Time written are metadata variables, but SNOW_PERSISTENCE can be plotted. Choose a colorscale and map overlay that you like.

![ncvis TS](../../../images/basics/Ncvis.clm6.snow_persistence.outlines.png)
*<p style="text-align: center;"> Figure: NCVis i1850sp.basics Instantaneous snow persistence Jan 01 0001 with the Haline colorscale and map outlines </p>*

By default, there is only one variable and one time slice written to the h0i file each month for this configuration. Later exercises will describe how to add more instantaneous or averaged output, more time slices per file, and other history file options.
</div>
