# Other Useful XML Variables 

| Variable | Description |
|---|---|
| `PIO_DEBUG_LEVEL` | Sets the level of standard output print statements. This variable is useful for debugging. If debugging, a higher value may be set | 
| `DOUT_S` | Turns on short-term archiving. `DOUT_S` is set to `TRUE` by default. |
| `HIST_OPTION` | Sets the driver snapshot history file frequency.  |
| `CCSM_CO2_PPMV` |  CO2 value to be propagated to CLM, when `OCN_CO2_TYPE` and `CLM_CO2_TYPE` are set to `'constant'`. |
